```python
from pathlib import Path
import pandas as pd
```

```python
# Carregar os arquivos CSV originais

compras = pd.read_csv("../data/raw/compras.csv")
itens = pd.read_csv("../data/raw/itens.csv")
resultados = pd.read_csv("../data/raw/resultados.csv")

# Exibir as dimensões de cada arquivo

for nome, df in [
    ("Compras", compras),
    ("Itens", itens),
    ("Resultados", resultados),
]:
    print(f"{nome}: {df.shape[0]:,} linhas e {df.shape[1]} colunas")
```

```python
for nome, df in [
    ("COMPRAS", compras),
    ("ITENS", itens),
    ("RESULTADOS", resultados),
]:
    print(f"\n{'=' * 60}")
    print(f"{nome}")
    print(f"{'=' * 60}")

    print("\nColunas e tipos:")
    display(
        df.dtypes.rename("tipo")
        .to_frame()
        .rename_axis("coluna")
    )

    print("\nPrimeiras 3 linhas:")
    display(df.head(5))
```

```python
# Diagnóstico inicial de qualidade dos dados

for nome, df in [
    ("COMPRAS", compras),
    ("ITENS", itens),
    ("RESULTADOS", resultados),
]:
    print(f"\n{'=' * 55}")
    print(f"{nome}")
    print(f"{'=' * 55}")

    # Quantidade de valores ausentes por coluna
    ausentes = df.isna().sum()
    ausentes = ausentes[ausentes > 0]

    print("\nValores ausentes:")
    if ausentes.empty:
        print("Nenhum valor ausente identificado.")
    else:
        display(ausentes.rename("quantidade").to_frame())

    # Quantidade de linhas totalmente duplicadas
    duplicadas = df.duplicated().sum()
    print(f"\nLinhas totalmente duplicadas: {duplicadas}")

    # Quantidade de valores distintos nas colunas
    print("\nQuantidade de valores distintos nas principais chaves:")

    chaves = [
        coluna for coluna in [
            "numero_controle_pncp",
            "numero_item",
            "sequencial_resultado",
            "cnpj_orgao",
            "ni_fornecedor",
        ]
        if coluna in df.columns
    ]

    display(df[chaves].nunique().rename("valores_distintos").to_frame())
```

```python
# 1. Verificar se as duas datas ausentes coincidem
datas_ausentes = (
    compras["data_abertura_proposta"].isna()
    & compras["data_encerramento_proposta"].isna()
)

print("Compras com ambas as datas ausentes:", datas_ausentes.sum())

print(
    "Compras com apenas uma das duas datas ausente:",
    (
        compras["data_abertura_proposta"].isna()
        != compras["data_encerramento_proposta"].isna()
    ).sum(),
)


# 2. Verificar duplicidade da chave de um item
chave_item = ["numero_controle_pncp", "numero_item"]

print(
    "\nItens com chave repetida:",
    itens.duplicated(subset=chave_item).sum()
)

# Verificar se a sequência do resultado se repete
chave_resultado = [
    "numero_controle_pncp",
    "numero_item",
    "sequencial_resultado",
]

print(
    "Resultados com chave completa repetida:",
    resultados.duplicated(subset=chave_resultado).sum()
)


# 3. Comparar as compras presentes nos arquivos
ids_compras = set(compras["numero_controle_pncp"])
ids_itens = set(itens["numero_controle_pncp"])
ids_resultados = set(resultados["numero_controle_pncp"])

print("\nCompras sem itens:", len(ids_compras - ids_itens))
print("Compras sem resultados:", len(ids_compras - ids_resultados))
print("Compras nos itens que não estão em compras:", len(ids_itens - ids_compras))
print(
    "Compras nos resultados que não estão em compras:",
    len(ids_resultados - ids_compras),
)


# 4. Conferir a indicação de resultado na tabela de itens
print("\nItens com indicação de resultado:")
display(itens["tem_resultado"].value_counts(dropna=False).to_frame("quantidade"))

print("\nSituação do item:")
display(itens["situacao_item"].value_counts(dropna=False).head(10).to_frame("quantidade"))
```

```python
## Verificar valores ausentes nas chaves candidatas - Valores "NaN"

compras = pd.read_csv("../data/raw/compras.csv")
itens = pd.read_csv("../data/raw/itens.csv")
resultados = pd.read_csv("../data/raw/resultados.csv")

# Chave candidata da tabela compras
print("COMPRAS")
print("Valores ausentes em numero_controle_pncp:",
      compras["numero_controle_pncp"].isna().sum())


# Chave candidata da tabela itens
print("\nITENS")
print("Valores ausentes em numero_controle_pncp:",
      itens["numero_controle_pncp"].isna().sum())
print("Valores ausentes em numero_item:",
      itens["numero_item"].isna().sum())


# Chave candidata da tabela resultados
print("\nRESULTADOS")
print("Valores ausentes em numero_controle_pncp:",
      resultados["numero_controle_pncp"].isna().sum())
print("Valores ausentes em numero_item:",
      resultados["numero_item"].isna().sum())
print("Valores ausentes em sequencial_resultado:",
      resultados["sequencial_resultado"].isna().sum())
```

```python
### Verificar valores ausentes ou vazios nas chaves candidatas - " "

# COMPRAS
print("COMPRAS")
print("numero_controle_pncp vazios:",
      compras["numero_controle_pncp"].astype("string").str.strip().eq("").sum())


# ITENS
print("\nITENS")
print("numero_controle_pncp vazios:",
      itens["numero_controle_pncp"].astype("string").str.strip().eq("").sum())
print("numero_item ausentes:",
      itens["numero_item"].isna().sum())


# RESULTADOS
print("\nRESULTADOS")
print("numero_controle_pncp vazios:",
      resultados["numero_controle_pncp"].astype("string").str.strip().eq("").sum())
print("numero_item ausentes:",
      resultados["numero_item"].isna().sum())
print("sequencial_resultado ausentes:",
      resultados["sequencial_resultado"].isna().sum())
```

### Diagnóstico inicial das chaves candidatas

Foram verificadas as seguintes chaves candidatas:

- **Compras:** `numero_controle_pncp`
- **Itens:** `numero_controle_pncp` + `numero_item`
- **Resultados:** `numero_controle_pncp` + `numero_item` + `sequencial_resultado`

As combinações não apresentaram duplicidades.
Não foram encontrados valores ausentes nas colunas das chaves, nem campos textuais vazios ou preenchidos apenas com espaços.

Portanto, não foram identificados problemas de unicidade ou preenchimento nas chaves candidatas durante o diagnóstico inicial.
